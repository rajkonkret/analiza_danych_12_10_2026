```python
# od pythona 3.10 match case

dane = "python", "java"

match dane:
    case "python":
        print("Znam pythona")
    case "java":
        print("Jave też znam")
    case "python", "java":
        print("Zanam wszytskie:)")
    case _:  # wartosc domyslna, odpiwednik else
        print("Nie znam")
```

# Klasy

Klasa - przepis, szablon
obiekt (instancja) - zbudowany wg przepisu

```python
class User:
    """Komentarz - dokumentacja"""

    def __init__(self, name, id):
        """Metoda inicjalizująca - (konstruktor)"""
        self.name = name
        self.id = id

    def powitanie(self):
        print("Nazywam się:", self.name)

    def __str__(self):
        return  f"{self.name} {self.id}"
```

```python
# cz1 = User()
cz1 = User("Radek", 1)
```

```python
print(cz1.__doc__)
```

```python
# pydoc
```

```python
print(cz1)  # po nadpisaniu __str__ -> Radek 1
```

```python
cz1.powitanie()
```
