# Klasy

Klasa - przepis, szablon
obiekt (instancja) - zbudowany wg przepisu

```python
class User:
    """Komentarz - dokumentacja"""

    def __init__(self, name, id):
        """Metoda inicjalizująca - (konstruktor)"""
        self.name = name
        self.id = id

    def powitanie(self):
        print("Nazywam się:", self.name)

    def __str__(self):
        return  f"{self.name} {self.id}"
```

```python
# cz1 = User()
cz1 = User("Radek", 1)
```

```python
print(cz1.__doc__)
```

```python
# pydoc
```

```python
print(cz1)  # po nadpisaniu __str__ -> Radek 1
```

```python
sprzedaz = np.array([120, 135, 98, 150, 180, 110, 95, 210, 170, 140])
```

```python
# wypisac podstawowe miary statystyczne, np.sum()
print("Mediana:", np.median(sprzedaz))
print("Minimum:", np.min(sprzedaz))
print("Maksimum:", np.max(sprzedaz))
print("Odchylenie:", np.std(sprzedaz))
print("Suma:", np.sum(sprzedaz))
```

```python
# duża sprzedaż - większa od średniej
duza_sprzedaz = sprzedaz[sprzedaz > np.mean(sprzedaz)]
print(duza_sprzedaz)
```

```python
# ile razy była duza sprzedaz
dni_duzej_sprzedazy = len(duza_sprzedaz)
dni_duzej_sprzedazy
```

```python
# dodac 10% do sprzedazy
# broadcasting
wzrost_sprzedazy = sprzedaz * 1.10
wzrost_sprzedazy
```

```python
duza_sprzedaz_mod = [liczba * 1.1 for liczba in duza_sprzedaz]
print(duza_sprzedaz_mod)
```

```python
# 3 sklepy, 4 miesiace

sprzedaz = np.array(
    [
        [120, 130, 140, 150],
        [100, 110, 115, 125],
        [200, 190,210, 220]
    ]
)
```

```python
print(sprzedaz)
```

```python
# oblicz:
# suma dla każdego sklepu
print(np.sum(sprzedaz[0]))
```

```python
suma_sklep = sprzedaz[0].sum()
suma_sklep
```

```python
suma_sklepy_wiersze = sprzedaz.sum(axis=1) # per sklep
print("Suma dla każdego sklepu:", suma_sklepy_wiersze)
```

```python
# suma dla kazdego miesiąca
suma_miesiace_kolumna = sprzedaz.sum(axis=0) # per miesiac
print("Suma dla każdego miesiąca:", suma_miesiace_kolumna)
```

```python
suma_wszystkie = sprzedaz.sum() # sumuje wszystkie wartości
print("Sumawszystkicha:", suma_wszystkie)
```

```python
najlepszy_sklep = np.argmax(suma_sklepy_wiersze) # poda index
print("Najlepszy sklep nr:", najlepszy_sklep + 1)
```
